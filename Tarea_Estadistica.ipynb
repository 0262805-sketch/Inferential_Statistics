### Suma de dos listas por posición

```python
L = [
    lista_a[0] + lista_b[0],
    lista_a[1] + lista_b[1],
    lista_a[2] + lista_b[2],
    lista_a[3] + lista_b[3],
    lista_a[4] + lista_b[4]
]

L
```

### Uso de for y range

```python
for i in range(5):
    print(i)
```

```python
L = [lista_a[i] + lista_b[i] for i in range(5)]
L
```

### Multiplicar valores de una lista

```python
2 * lista_b
```

```python
0.5 * lista_b
```

```python
M = [0.5 * lista_b[i] for i in range(5)]
M
```

### Conversión de arreglos a matrices

```python
valores = [i for i in range(10, 30)]
valores
```

```python
A = np.array(valores)
A
```

```python
matrix_a = A.reshape(5, 4)
matrix_a
```

```python
matrix_b = A.reshape(10, 2)
matrix_b
```

```python
matriz3 = A.reshape(2, 10)
matriz3
```
